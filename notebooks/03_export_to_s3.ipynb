```python
OUTPUT_BUCKET_PATH = "s3://mortgage-risk-api/api"

TABLE_NAMES = [
    "core_portfolio_findings",
    "final_business_findings",
    "state_findings",
    "segment_findings",
    "vintage_findings",
    "early_prepay_findings",
    "transition_30d_watchlist",
    "distress_persistence_watchlist",
    "credit_risk_ladder",
    "ltv_risk_ladder",
    "dti_risk_ladder",
]

SOURCE_PREFIX = "mortgage_risk.analytics"

for table_name in TABLE_NAMES:
    source_table = f"{SOURCE_PREFIX}.{table_name}"
    output_path = f"{OUTPUT_BUCKET_PATH}/{table_name}"

    df = spark.table(source_table)

    (
        df.write
        .mode("overwrite")
        .parquet(output_path)
    )

    print(f"Saved : {source_table} -> {output_path}")
```
